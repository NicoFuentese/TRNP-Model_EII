```python
import numpy as np
import matplotlib.pyplot as plt

v_f = 60  # km/h
k_j = 200 # veh/km

k = np.linspace(0, k_j, 400)
q = v_f * (k - k**2 / k_j)

k_c = k_j / 2
q_c = v_f * (k_c - k_c**2 / k_j)

plt.figure()
plt.plot(k, q, color="darkblue")
plt.scatter([k_c], [q_c], color="darkblue")
plt.axvline(k_c, linestyle='--', color="darkred")
plt.axvline(k_j, linestyle=':', color="indigo")
plt.text(k_c, q_c, "  $k_c$", verticalalignment='bottom')
plt.text(k_j, 0, "  $k_j$", verticalalignment='bottom')
plt.xlabel("Densidad $k$ (veh/km)")
plt.ylabel("Flujo $q$ (veh/h)")
plt.title("Curva flujo–densidad en el modelo de Greenshields")
plt.grid()
plt.tight_layout()
plt.show()

"""
plot_path = "/content/greenshields_qk.png"
plt.savefig(plot_path)
plot_path
"""
```
